whether the cancerous cell is malignant or benign

StandardScaler is preprocessing technique
under standardScaler there is linear regression, under that there is logistic regression, we have taken gaussian naive bayes, we have imported decision tree classification, we will comapre different models. pandas because we are working with dataset

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,confusion_matrix)
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = load_breast_cancer()
X = data.data
y = data.target
print("Dataset loaded successful")
print("Number of samples", X.shape[0])
print("Number of features", X.shape[1])

print ("\nClasses:")
print(data.target_names)
```

```python
# split the dataset

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42, # starting from any serial no
    stratify=y # we will divide y as if benign and malignant are in same proportion and to avoid bias
)
print("\nTraining samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
#feature scaling
#we will be using standardscaler for x only, y is there "malignant and benign"

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
#creating four models

logistic_model = LogisticRegression(max_iter=1000)
knn_model = KNeighborsClassifier(n_neighbors=5)
naive_bayes_model = GaussianNB()
decision_tree_model = DecisionTreeClassifier(
    random_state=42
)
```

```python
#logistic Regression

logistic_model.fit(X_train_scaled, y_train)
logistic_pred = logistic_model.predict(X_test_scaled)

logistic_accuracy = accuracy_score(
    y_test,
    logistic_pred
)

logistic_precision = precision_score(
    y_test,
    logistic_pred
)

logistic_recall = recall_score(
    y_test,
    logistic_pred
)

logistic_f1 = f1_score(
    y_test,
    logistic_pred
)

print("Accuracy:",round(logistic_acuracy, 4))
print("precision:",round(logistic_precision, 4))
print("recall:",round(logistic_recall, 4))
print("f1:",round(logistic_f1, 4))

print("\nConfussion Matrix")
print(
    confusion_matrix(
        y_test,
        logistic_pred
    )
)
```

```python
#knn

knn_model.fit(X_train_scaled, y_train)
knn_pred = knn_model.predict(X_test_scaled)

knn_accuracy = accuracy_score(
    y_test,
    knn_pred
)

knn_precision = precision_score(
    y_test,
    knn_pred
)

knn_recall = recall_score(
    y_test,
    knn_pred
)

knn_f1 = f1_score(
    y_test,
    knn_pred
)

print("Accuracy:",round(knn_acuracy, 4))
print("precision:",round(knn_precision, 4))
print("recall:",round(knn_recall, 4))
print("f1:",round(knn_f1, 4))

print("\nConfussion Matrix")
print(
    confusion_matrix(
        y_test,
        knn_pred
    )
)
```

```python
#naive bayes

naive_bayes_model.fit(X_train_scaled, y_train)
naive_bayes_pred = knn_model.predict(X_test_scaled)

naive_bayes_accuracy = accuracy_score(
    y_test,
    naive_bayes_pred
)

naive_bayes_precision = precision_score(
    y_test,
    naive_bayes_pred
)

naive_bayes_recall = recall_score(
    y_test,
    naive_bayes_pred
)

naive_bayes_f1 = f1_score(
    y_test,
    naive_bayes_pred
)

print("Accuracy:",round(naive_bayes_acuracy, 4))
print("precision:",round(naive_bayes_precision, 4))
print("recall:",round(naive_bayes_recall, 4))
print("f1:",round(naive_bayes_f1, 4))

print("\nConfussion Matrix")
print(
    confusion_matrix(
        y_test,
        naive_bayes_pred
    )
)
```

```python
#decision tree

decision_tree_model.fit(X_train, y_train)
decision_tree_pred = decision_tree_model.predict(X_test)

decision_tree_accuracy = accuracy_score(
    y_test,
    decision_tree_pred
)

decision_tree_precision = precision_score(
    y_test,
    decision_tree_pred
)

decision_tree_recall = recall_score(
    y_test,
    decision_tree_pred
)

decision_tree_f1 = f1_score(
    y_test,
    decision_tree_pred
)

print("Accuracy:",round(decision_tree_acuracy, 4))
print("precision:",round(decision_tree_precision, 4))
print("recall:",round(decision_tree_recall, 4))
print("f1:",round(decision_tree_f1, 4))

print("\nConfussion Matrix")
print(
    confusion_matrix(
        y_test,
        decision_tree_pred
    )
)
```

```python
results = {

    "Algorith": [
        "Logistic Regression",
        "KNN",
        "Naive Bayes",
        "Decision Tree"
    ],

    "Accuracy": [
        logistic_accuracy,
        knn_accuracy,
        naive_bayes_accuracy,
        decision_tree_accuracy
    ],

    "Precision": [
        logistic_precision,
        knn_precision,
        naive_bayes_precision,
        decision_tree_precision
    ],

    "Recall": [
        logistic_recall,
        knn_recall,
        naive_bayes_recall,
        decision_tree_recall
    ],

    "f1_score": [
        logistic_f1,
        knn_f1,
        naive_bayes_f1,
        decision_tree_f1
    ],

}
results_df = pd.DataFrame(results)
```

```python
print("\n")
print("=" * 75)
print("PERFORMANCE COMPARISION")
print("=" * 75)

print(
    results_df.to_string(
        index=False
    )
)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Reshape results_df for easier plotting with seaborn
df_melted = results_df.melt(id_vars='Algorith', var_name='Metric', value_name='Value')

# Create the plot
plt.figure(figsize=(12, 7))
sns.set_theme(style="whitegrid")

ax = sns.barplot(
    data=df_melted,
    x='Algorith',
    y='Value',
    hue='Metric',
    palette='viridis'
)

# Customize chart labels and title
plt.title('Model Performance Comparison', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Algorithms', fontsize=12, labelpad=10)
plt.ylabel('Score Value', fontsize=12, labelpad=10)
plt.ylim(0.8, 1.02)  # Adjusting limit to highlight differences in high performance
plt.legend(title='Metrics', bbox_to_anchor=(1.05, 1), loc='upper left')

# Add values on top of the bars
for container in ax.containers:
    ax.bar_label(container, fmt='%.3f', padding=3, fontsize=9)

plt.tight_layout()
plt.show()
```
